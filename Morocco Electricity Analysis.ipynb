#Task 5: Data Visualisation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

zone_cols = [
    'PowerConsumption_Zone1',
    'PowerConsumption_Zone2',
    'PowerConsumption_Zone3'
]

environment_cols = ['Temperature', 'Humidity', 'WindSpeed']

# Ensure the date column is in datetime format
df['Datetime'] = pd.to_datetime(df['Datetime'])

# Create Date if it is not already available
if 'Date' not in df.columns:
    df['Date'] = df['Datetime'].dt.date

# Create time fields if they are not already available
if 'Hour' not in df.columns:
    df['Hour'] = df['Datetime'].dt.hour

if 'Month' not in df.columns:
    df['Month'] = df['Datetime'].dt.month

if 'Day_of_Week' not in df.columns:
    df['Day_of_Week'] = df['Datetime'].dt.day_name()
```

```python
#1. Electricity consumption trends over time
daily_trend = df.groupby('Date')[zone_cols].mean().reset_index()
daily_trend['Date'] = pd.to_datetime(daily_trend['Date'])

daily_long = daily_trend.melt(
    id_vars='Date',
    value_vars=zone_cols,
    var_name='Zone',
    value_name='Average Consumption'
)

fig = px.line(
    daily_long,
    x='Date',
    y='Average Consumption',
    color='Zone',
    title='Daily Electricity Consumption Trends Across Three Zones',
    labels={'Date': 'Date', 'Average Consumption': 'Average Electricity Consumption'}
)

fig.update_layout(
    template='plotly_white',
    hovermode='x unified',
    legend_title='Consumption Zone'
)

fig.show()
```

```python
#2. Average consumption by zone
zone_averages = df[zone_cols].mean().reset_index()
zone_averages.columns = ['Zone', 'Average Consumption']

fig = px.bar(
    zone_averages,
    x='Zone',
    y='Average Consumption',
    title='Average Electricity Consumption by Zone',
    text_auto='.2s',
    labels={'Average Consumption': 'Average Electricity Consumption'}
)

fig.update_layout(template='plotly_white')
fig.show()
```

```python
#3. Hourly consumption patterns
hourly_average = df.groupby('Hour')[zone_cols].mean().reset_index()

hourly_long = hourly_average.melt(
    id_vars='Hour',
    value_vars=zone_cols,
    var_name='Zone',
    value_name='Average Consumption'
)

fig = px.line(
    hourly_long,
    x='Hour',
    y='Average Consumption',
    color='Zone',
    markers=True,
    title='Average Electricity Consumption by Hour',
    labels={'Hour': 'Hour of the Day', 'Average Consumption': 'Average Electricity Consumption'}
)

fig.update_layout(
    template='plotly_white',
    xaxis=dict(dtick=1),
    legend_title='Consumption Zone'
)

fig.show()
```

```python
#4. Monthly consumption patterns
monthly_average = df.groupby('Month')[zone_cols].mean().reset_index()

monthly_long = monthly_average.melt(
    id_vars='Month',
    value_vars=zone_cols,
    var_name='Zone',
    value_name='Average Consumption'
)

month_names = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}
monthly_long['Month Name'] = monthly_long['Month'].map(month_names)

fig = px.line(
    monthly_long,
    x='Month Name',
    y='Average Consumption',
    color='Zone',
    markers=True,
    category_orders={'Month Name': list(month_names.values())},
    title='Average Electricity Consumption by Month',
    labels={'Month Name': 'Month', 'Average Consumption': 'Average Electricity Consumption'}
)

fig.update_layout(template='plotly_white', legend_title='Consumption Zone')
fig.show()
```

```python
#5. Day-of-week comparisons
weekday_order = [
    'Monday', 'Tuesday', 'Wednesday', 'Thursday',
    'Friday', 'Saturday', 'Sunday'
]

weekday_average = df.groupby('Day_of_Week')[zone_cols].mean().reindex(weekday_order).reset_index()

weekday_long = weekday_average.melt(
    id_vars='Day_of_Week',
    value_vars=zone_cols,
    var_name='Zone',
    value_name='Average Consumption'
)

fig = px.bar(
    weekday_long,
    x='Day_of_Week',
    y='Average Consumption',
    color='Zone',
    barmode='group',
    category_orders={'Day_of_Week': weekday_order},
    title='Average Electricity Consumption by Day of the Week',
    labels={'Day_of_Week': 'Day of the Week',
            'Average Consumption': 'Average Electricity Consumption'}
)

fig.update_layout(template='plotly_white', legend_title='Consumption Zone')
fig.show()
```

```python
#6. Environmental factor relationships
# Pearson correlation between environmental factors and the three zones
correlation_results = df[environment_cols + zone_cols].corr(method='pearson')
environment_zone_corr = correlation_results.loc[environment_cols, zone_cols]

plt.figure(figsize=(10, 6))
sns.heatmap(
    environment_zone_corr,
    annot=True,
    cmap='coolwarm',
    center=0,
    fmt='.2f'
)
plt.title('Environmental Factors and Electricity Consumption')
plt.xlabel('Electricity Consumption Zones')
plt.ylabel('Environmental Factors')
plt.tight_layout()
plt.show()

# Scatter plots using a sample to keep the charts readable
sample_df = df.sample(n=min(5000, len(df)), random_state=42)

for factor in environment_cols:
    for zone in zone_cols:
        plt.figure(figsize=(7, 5))
        sns.scatterplot(
            data=sample_df,
            x=factor,
            y=zone,
            alpha=0.3
        )
        plt.title(f'{factor} vs {zone}')
        plt.xlabel(factor)
        plt.ylabel('Electricity Consumption')
        plt.tight_layout()
        plt.show()
```

```python
#7. Comparison of the three zones

zone_long = df[zone_cols].melt(
    var_name='Zone',
    value_name='Electricity Consumption'
)

fig = px.box(
    zone_long,
    x='Zone',
    y='Electricity Consumption',
    title='Distribution of Electricity Consumption Across the Three Zones',
    points=False,
    labels={'Electricity Consumption': 'Electricity Consumption'}
)

fig.update_layout(template='plotly_white')
fig.show()
```

#Task 6: Business Insights and Recommendations

#Key Findings

Based on the analysis of electricity consumption across the three zones, the following findings were identified:

1. Zone 1 records the highest average consumption.
Zone 1 has an average consumption reading of approximately 32,344.97, compared with 21,042.51 for Zone 2 and 17,835.41 for Zone 3. This indicates that 
Zone 1 has the highest average electricity demand among the three zones. The organisation should therefore pay particular attention to resource 
allocation and demand management in this area.

2. Zone 3 records the lowest average consumption.
Zone 3 has the lowest average consumption reading of approximately 17,835.41. This suggests that its electricity demand is lower than that of the other
two zones. The organisation can use this information when comparing demand levels and planning resources across geographical areas.

3. Electricity consumption is generally highest during the evening.
The hourly analysis shows that consumption across the three zones tends to reach its highest levels around 20:00, while the lowest readings generally 
occur between 06:00 and 07:00. This suggests that electricity demand is higher during evening hours than early morning hours.

4. Consumption varies across the months.
The monthly analysis shows differences in consumption patterns across the three zones. Zone 1 and Zone 2 record their highest average readings in 
August, while Zone 3 records its highest in July. The lowest average readings occur in November for Zone 1, April for Zone 2 and December for Zone 3. 
These differences indicate that demand patterns are not identical across the zones.

5. Consumption differs across the days of the week.
The day-of-week analysis shows that Zone 1 records its highest average consumption on Wednesday, while Zone 2 records its highest on Thursday. Both 
zones record their lowest average consumption on Sunday. This suggests that electricity demand changes depending on the day of the week.

6. The three zones have different levels of consumption variability.
Zone 1 has the highest standard deviation of approximately 7,130.56, followed by Zone 3 at 6,622.17 and Zone 2 at 5,201.47. This indicates that Zone 1 
has the greatest variation in its consumption readings, while Zone 2 has the most stable readings among the three zones.

7. Environmental Factors and Electricity Consumption
Temperature had a positive relationship with electricity consumption across all three zones, with the strongest relationship observed in Zone 3 
(Pearson correlation: 0.490). Humidity showed a negative relationship with consumption across the zones, while wind speed had weak positive relationships. 
The Spearman results showed similar patterns. These findings suggest that temperature and humidity may be useful factors to consider when analysing 
electricity demand. However, correlation does not establish cause and effect.


#Business Implications

The findings provide useful information for an energy management organisation responsible for monitoring electricity demand. The higher average 
consumption in Zone 1 indicates that this zone may require greater attention when planning electricity supply and managing demand. 
The evening peak suggests that the organisation should prepare for increased demand around 20:00. Differences in monthly and weekly consumption 
patterns also show the importance of considering both seasonal and day-to-day variations when planning electricity resources. Furthermore, 
the differences in variability between the zones suggest that each geographical area may require a different demand management approach.

#Recommendations

Based on the findings, the following recommendations are proposed:
-Strengthen demand management in Zone 1: The organisation should closely monitor Zone 1 because it records the highest average consumption and greatest
variability.

-Prepare for evening demand: Electricity supply and demand management plans should consider the higher consumption readings recorded around 20:00 to 
reduce the risk of supply pressure during peak hours.

-Use historical patterns for resource planning: The organisation should consider monthly and weekly consumption patterns when planning electricity 
distribution and allocating resources across the three zones.

-Monitor each zone separately: Since the three zones have different consumption levels and variability, the organisation should use zone-specific monitoring and planning strategies rather than assuming that demand patterns are identical.

#Conclusion
The analysis identified differences in electricity consumption across the three geographical zones, including variations in average consumption, hourly
patterns, monthly trends and weekly behaviour. Zone 1 recorded the highest average consumption and greatest variability, while Zone 3 recorded the 
lowest average consumption. These findings can support better electricity demand monitoring, resource allocation and planning. The analysis covers one 
year of observations, so further monitoring over additional years would help establish whether the identified patterns continue over time.
